```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
miss_mod=False
root = "/data/edu/"
join = os.path.join
train = join(root, r"train_final.csv") if not miss_mod else join(root, r"train_miss.csv")
val = join(root, r"val_final.csv") if not miss_mod else join(root, r"val_miss.csv")
tes = join(root, r"test_final.csv") if not miss_mod else join(root, r"test_miss.csv")
train = pd.read_csv(train)
val = pd.read_csv(val)
tes = pd.read_csv(tes)
```

```python
total = pd.concat([train,val,tes])
## Brain Stats
brain_ds = total[total["Type"]==1]
brain_dis = [len(brain_ds[brain_ds["Result"]==i]) for i in pd.unique(brain_ds["Result"])]
func = lambda x: x.split("/")[2]
lab=pd.unique([ "No Tumor", "Glioma", "Meningioma","Pituitary"])
_,_,autopcts=plt.pie(brain_dis, labels=lab,  autopct='%1.1f%%', textprops={"size":12.5})
plt.setp(autopcts, **{'fontsize':12.5})
plt.show()

## Liver Stats
liver_ds = total[total["Type"]==0]
liver_dis = [len(liver_ds[liver_ds["Result"]==i]) for i in pd.unique(liver_ds["Result"])]
lab=["Hepatocellular Carcinoma ","Cholangiocarcinoma"]
plt.pie(liver_dis, labels=lab,  autopct='%1.1f%%', textprops={"size":12.5})
plt.show()


## Eye Stats
glaucoma_ds = total[total["Type"]==2]
glaucoma_dis = [len(glaucoma_ds[glaucoma_ds["Result"]==i]) for i in pd.unique(glaucoma_ds["Result"])]
lab=["Glaucoma ","Non Glaucoma"]
plt.pie(glaucoma_dis, labels=lab,  autopct='%1.1f%%', textprops={"size":12.5})
plt.show()

print([len(total[total["Type"]==0]), len(total[total["Type"]==1]), len(total[total["Type"]==2])])
plt.bar(["Tumor Dataset", "HCC & GCC Dataset", "Glaucoma Dataset"],[len(total[total["Type"]==0]), len(total[total["Type"]==1]), len(total[total["Type"]==2])])
```
